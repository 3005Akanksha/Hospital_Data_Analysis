```python
for name, df in sheets.items():

    print(
        name,
        "Duplicate Rows:",
        df.duplicated().sum()
    )
```

```python
for name in sheets:

    sheets[name] = sheets[name].drop_duplicates()

print("Duplicate checking completed!")
```

```python
primary_keys = {
    'Patients': 'PatientID',
    'EDVisits': 'EDVisitID',
    'AmbulatoryVisits': 'VisitID',
    'ReAdmissionRegistry': 'AdmissionID',
    'Discharges': 'AdmissionID',
    'Providers': 'ProviderID',
    'EDUnique': 'EDVisitID'
}

for name, key in primary_keys.items():

    df = sheets[name]

    print(
        name,
        "Duplicate IDs:",
        df[key].duplicated().sum()
    )
```

```python
for name in sheets:

    df = sheets[name].copy()

    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(r'[^a-zA-Z0-9]+', '_', regex=True)
        .str.strip('_')
    )

    sheets[name] = df

print("Column names standardized!")
```

```python
for name in sheets:

    df = sheets[name].copy()

    for col in df.select_dtypes(include='object').columns:

        df[col] = df[col].apply(
            lambda x: x.strip() if isinstance(x, str) else x
        )

    sheets[name] = df

print("Text cleaning completed!")
```

```python
for name in ['EDVisits', 'EDUnique']:

    sheets[name]['eddisposition'] = (
        sheets[name]['eddisposition']
        .astype('string')
        .str.strip()
        .str.title()
    )

print("ED Disposition cleaned successfully!")
```

```python
date_columns = {
    'Patients': ['dateofbirth'],
    'EDVisits': ['visitdate', 'eddischargetime'],
    'AmbulatoryVisits': ['dateofvisit', 'datescheduled'],
    'ReAdmissionRegistry': ['admissiondate', 'dischargedate'],
    'Discharges': ['admissiondate', 'dischargedate'],
    'Providers': ['providerdateonstaff'],
    'EDUnique': ['visitdate', 'eddischargetime']
}

for name, cols in date_columns.items():

    for col in cols:

        if col in sheets[name].columns:

            sheets[name][col] = pd.to_datetime(
                sheets[name][col],
                errors='coerce'
            )

print("Date conversion completed successfully!")
```

```python
print("Emergency Acuity:")
display(sheets['EDVisits']['acuity'].value_counts(dropna=False))

print("Ambulatory Vital Signs:")
display(
    sheets['AmbulatoryVisits'][
        ['bloodpressuresystolic',
         'bloodpressurediastolic',
         'pulse']
    ].describe()
)

print("Expected Mortality:")
display(sheets['ReAdmissionRegistry']['expectedmortality'].describe())

print("Expected LOS:")
display(sheets['ReAdmissionRegistry']['expectedlos'].describe())
```

```python
readmission = sheets['ReAdmissionRegistry'].copy()

readmission['actual_los_days'] = (
    readmission['dischargedate'] -
    readmission['admissiondate']
).dt.total_seconds() / 86400

readmission['actual_los_days'] = (
    readmission['actual_los_days'].round(2)
)

display(readmission.head())
```

```python
sheets['ReAdmissionRegistry'] = readmission
```

```python
def los_category(days):

    if pd.isna(days):
        return 'Unknown'

    elif days <= 3:
        return 'Short Stay'

    elif days <= 7:
        return 'Medium Stay'

    else:
        return 'Long Stay'


readmission['los_category'] = (
    readmission['actual_los_days']
    .apply(los_category)
)

sheets['ReAdmissionRegistry'] = readmission

display(readmission[['actual_los_days', 'los_category']].head(10))
```

```python
patients = sheets['Patients'].copy()
ed_visits = sheets['EDVisits'].copy()

ed_visits = ed_visits.merge(
    patients[['patientid', 'dateofbirth']],
    on='patientid',
    how='left',
    validate='many_to_one'
)

ed_visits['patient_age'] = (
    ed_visits['visitdate'] -
    ed_visits['dateofbirth']
).dt.days / 365.2425

ed_visits['patient_age'] = (
    ed_visits['patient_age'].round(1)
)

sheets['EDVisits'] = ed_visits

display(ed_visits[['patientid', 'patient_age']].head())
```

```python
ed_visits['age_group'] = pd.cut(
    ed_visits['patient_age'],
    bins=[0, 18, 35, 50, 65, 120],
    labels=[
        '0-18',
        '19-35',
        '36-50',
        '51-65',
        '65+'
    ],
    include_lowest=True
)

sheets['EDVisits'] = ed_visits

display(
    ed_visits[['patient_age', 'age_group']].head(10)
)
```

```python
missing_summary = []

for name, df in sheets.items():
    for col in df.columns:
        missing_count = df[col].isna().sum()

        if missing_count > 0:
            missing_summary.append({
                'Sheet': name,
                'Column': col,
                'Missing Values': missing_count,
                'Missing %': round(
                    (missing_count / len(df)) * 100, 2
                )
            })

missing_df = pd.DataFrame(missing_summary)

if missing_df.empty:
    print("No missing values found!")
else:
    display(missing_df)
```

```python
key_columns = {
    'Patients': ['patientid'],
    'EDVisits': ['edvisitid', 'patientid'],
    'AmbulatoryVisits': ['visitid', 'patientid', 'providerid'],
    'ReAdmissionRegistry': ['admissionid', 'patientid'],
    'Discharges': ['admissionid', 'patientid'],
    'Providers': ['providerid'],
    'EDUnique': ['patientid']
}

for sheet_name, columns in key_columns.items():
    df = sheets[sheet_name]

    print(f"\n--- {sheet_name} ---")

    for col in columns:
        if col in df.columns:
            print(
                f"{col}: {df[col].nunique()} unique values "
                f"/ {df[col].isna().sum()} missing"
            )
        else:
            print(f"{col}: NOT FOUND")
```

```python
# Patient IDs
patient_ids = set(sheets['Patients']['patientid'].dropna())

tables_to_check = [
    'EDVisits',
    'AmbulatoryVisits',
    'ReAdmissionRegistry',
    'Discharges'
]

for name in tables_to_check:
    df = sheets[name]

    if 'patientid' in df.columns:
        unmatched = (
            df['patientid'].notna() &
            ~df['patientid'].isin(patient_ids)
        ).sum()

        print(f"{name}: {unmatched} unmatched Patient IDs")
```

```python
output_file = 'HospitalDatabase_Cleaned.xlsx'

with pd.ExcelWriter(output_file, engine='openpyxl') as writer:
    for name, df in sheets.items():
        df.to_excel(
            writer,
            sheet_name=name,
            index=False
        )

print("Cleaned Excel workbook created successfully!")
```
